I. Análisis de Eficiencia Algorítmica y Sobriedad Digital
> Realizar un código eficiente permite un menor consumo energético.
En este sentido, la lógica es "Hacer menos, pero mejor".

```python
# Importamos Librerías
import time # Para medir duración del código
import matplotlib.pyplot as plt # Para graficar
```

```python
# Algoritmo Base, con todos los divisores desde 1 hasta n.
# Definir función para que la tarea se repita
def es_primo(n):
    if n < 1:
        return False

    divisores = 0
    for i in range(1, n + 1): # Bucle
        if n % i == 0: # El residuo de la división debe ser cero
            divisores += 1

    if divisores == 2:
        return True
    else:
        return False
```

```python
# Algorimo Optimizado
def es_primo_o(n):
    if n <= 1:
        return False

    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False

    return True
```

```python
# Diferencia de ambos rendimientos
numero = 526000000

inicio_ab = time.time() # AB: Algoritmo Base
es_primo(numero)
fin_ab = time.time()
tiempo_ab = fin_ab - inicio_ab

inicio_o = time.time() # O: Algoritmo Optimizado
es_primo_o(numero)
fin_o = time.time()
tiempo_o = fin_o - inicio_o

print(f"Tiempo de ejecución del algoritmo base: {tiempo_ab}")
print(f"Tiempo de ejecución del algoritmo optimizado: {tiempo_o}")
```

```python
# Gráfico
etiquetas = ["Algoritmo Base", "Algoritmo Optimizado"]
tiempos = [tiempo_ab, tiempo_o]

plt.figure(figsize=(8,5))
plt.bar(etiquetas, tiempos, color=['#c1121f','#003049'])
plt.title('Comparación de Tiempos de Ejecución')
```
